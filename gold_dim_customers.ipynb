```python
import pyspark.sql.functions as F
# from pyspark.sql.windows import Window
```

# Reading from Silver Table

```python
query = """
SELECT 
ROW_NUMBER() OVER(ORDER BY ci.customer_id) AS customer_key,
ci.customer_id,
ci.first_name,
ci.last_name,
ci.marital_status,
ci.gender,
ci.created_date
FROM silver.crm_customers ci
"""
df = spark.sql(query)
```

#Business Transfornation and Data Modeling

# Write it to Gold Table

```python
(
df.write
.mode("overwrite")
.format("delta")
.saveAsTable("gold.dim_customers")
)
```
